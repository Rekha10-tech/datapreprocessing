```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
numerical_columns = [
    'Year',
    'Mileage_Km',
    'Engine_CC',
    'Power_BHP',
    'Previous_Owners',
    'Accidents_Reported',
    'Service_Score'
]

categorical_columns = [
    'Brand',
    'Fuel_Type',
    'Transmission',
    'City',
    'Seller_Type',
    'Condition'
]

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
X = df.drop(columns=['Resale_Price_Lakh', 'Car_ID'])
y = df['Resale_Price_Lakh']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
outlier_columns = [
    'Mileage_Km',
    'Engine_CC',
    'Power_BHP'
]

for column in outlier_columns:

    Q1 = X_train[column].quantile(0.25)
    Q3 = X_train[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = X_train[
        (X_train[column] < lower_bound) |
        (X_train[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of outliers:", len(outliers))
```

```python
X_train_clean = X_train.copy()
X_test_clean = X_test.copy()

iqr_bounds = {}

for column in outlier_columns:

    Q1 = X_train[column].quantile(0.25)
    Q3 = X_train[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    iqr_bounds[column] = (lower_bound, upper_bound)

    # Cap training data
    X_train_clean[column] = X_train_clean[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )

    # Apply the same training bounds to test data
    X_test_clean[column] = X_test_clean[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outliers handled using IQR-based capping.")
```

```python
for column in outlier_columns:

    lower_bound, upper_bound = iqr_bounds[column]

    remaining_outliers = X_train_clean[
        (X_train_clean[column] < lower_bound) |
        (X_train_clean[column] > upper_bound)
    ]

    print(
        column,
        "→ Remaining outliers:",
        len(remaining_outliers)
    )
```

```python
encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)
```

```python
scaler = StandardScaler()
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            'numerical',
            StandardScaler(),
            numerical_columns
        ),
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_columns
        )
    ]
)

print("Preprocessing transformation created.")
```

```python
X_train_processed = preprocessor.fit_transform(X_train_clean)
```

```python
X_test_processed = preprocessor.transform(X_test_clean)
```

```python
# Training data
preprocessor.fit_transform(X_train_clean)

# Testing data
preprocessor.transform(X_test_clean)
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Number of features after preprocessing:",
      len(feature_names))

print("\nProcessed feature names:")
print(feature_names)
```

```python
X_train_processed_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)

print("Processed training data:")
display(X_train_processed_df.head())
```

```python
print(
    "Missing values in training data:",
    X_train_processed_df.isnull().sum().sum()
)

print(
    "Missing values in testing data:",
    X_test_processed_df.isnull().sum().sum()
)
```

```python
scaled_columns = [
    column for column in X_train_processed_df.columns
    if column.startswith('numerical__')
]

print("Mean of scaled numerical features:")
print(
    X_train_processed_df[scaled_columns]
    .mean()
    .round(2)
)

print("\nStandard deviation of scaled numerical features:")
print(
    X_train_processed_df[scaled_columns]
    .std()
    .round(2)
)
```

```python
train_processed = X_train_processed_df.copy()
train_processed['Resale_Price_Lakh'] = y_train

test_processed = X_test_processed_df.copy()
test_processed['Resale_Price_Lakh'] = y_test

print("Final training dataset:")
display(train_processed.head())
```

```python
print("Original dataset shape:", df.shape)

print("\nFinal training dataset shape:",
      train_processed.shape)

print("Final testing dataset shape:",
      test_processed.shape)

print("\nMissing values in final training dataset:",
      train_processed.isnull().sum().sum())

print("Missing values in final testing dataset:",
      test_processed.isnull().sum().sum())
```

```python
train_processed.to_csv(
    'Used_Car_Preprocessed_Train.csv',
    index=False
)

test_processed.to_csv(
    'Used_Car_Preprocessed_Test.csv',
    index=False
)

print("Preprocessed datasets saved successfully.")
```

## Preprocessing Decisions

- The dataset was first inspected using `head()`, `info()`, and `describe()`.

- Missing values were checked using `isnull()`. No missing values were found, so missing-value imputation was not required.

- Duplicate records were checked using `duplicated()` and removed if present.

- `Car_ID` was removed because it is an identifier and is not a useful predictive feature.

- `Resale_Price_Lakh` was selected as the target variable and the remaining relevant columns were used as features.

- The data was divided into 80% training data and 20% testing data.

- Outliers were identified using the IQR method for continuous numerical variables. IQR-based capping was used to handle extreme values without unnecessarily deleting observations.

- Categorical variables were encoded using One-Hot Encoding because they represent nominal categories.

- Numerical features were standardized using StandardScaler.

- The preprocessing transformations were fitted only on the training data and then applied to the testing data. This prevents data leakage.

- The processed datasets were verified for missing values, correct dimensions, encoding, and scaling.

- The final preprocessed training and testing datasets were saved as CSV files.
